                                                     AUTOMATED DATA CLEANING & ANALYTICS ENGINE

```python
import numpy as np 
import pandas as pd
class DataLoader:
    """ Class responsible for loading and initial inspection of dataset."""
    def __init__(self, file_path):
        self.file_path = file_path
        self.df = None
    def load_csv(self):
        try:
            self.df = pd.read_csv(self.file_path)
            print(f"[SUCCESS] File loaded succesfully from: {self.file_path}")
            return self.df
        except Exception as e:
            print(f"[ERROR] Failed to load file: {e}")
            return None
class DataCleaner:
    """Class containing core Object-oriented methods to clean datasets."""
    def __init__(self, dataframe):
        self.df = dataframe
    def remove_duplicates(self):
        initial_rows = len(self.df)
        self.df = self.df.drop_duplicates()
        remove_count = initial_rows - len(self.df)
        print(f"[CLEANING] Removed {remove_count} duplicates rows(s).")
        return self.df
    def handle_missing_values(self):
        num_cols = self.df.select_dtypes(include = [np.number]).columns
        for col in num_cols:
            if self.df[col].isnull().sum()>0:
                median_val = self.df[col].median()
                self.df[col]=self.df[col].fillna(median_val)
                print(f"[CLEANING] Filling missing values in numerical column '{col}' with Median ({median_val}).")
        #Categorical column
        cat_cols = self.df.select_dtypes(include = ['object', 'string']).columns
        for col in cat_cols:
            if self.df[col].isnull().sum()>0:
                self.df[col].fillna('Unknown', inplace= True)
                print(f"[CLEANING] Failed missing values in categorical column '{col}' with 'Unknown'.")
        return self.df
class DataReporter:
    """Class responsible for generating analytics and summary reports."""
    def __init__(self, dataframe):
        self.df = dataframe
    def generate_summary(self):
        print("\n" + "="*40)
        print("             Data Health Report            ")
        print("="*40)
        print(f"Total Rows: {self.df.shape[0]}")
        print("\n---Missing Values Check ---")
        print(self.df.isnull().sum())
        print("\n---Numerical Summary Statistics---")
        print(self.df.describe())
        print("="*40 + "\n")
#DRIVER CODE
if __name__ == "__main__":
    raw_data = {
        "Age":[25,30, np.nan, 35, 30, 40],
        "salary":[50000, 60000, 700000, 55000, np.nan, 60000],
        "Department":["IT", "HR", "IT", "FINANCE", "HR", None]
    }
    sample_df = pd.DataFrame(raw_data)
    sample_df.to_csv("sample_unclean_data.csv", index = False)
    print("--- Starting pipeline Execution ---\n")
    loader = DataLoader("sample_unclean_data.csv")
    df = loader.load_csv()
    #clean data
    cleaner = DataCleaner(df)
    cleaner.remove_duplicates()
    clean_df = cleaner.handle_missing_values()
    #report output
    reporter = DataReporter(clean_df)
    reporter.generate_summary()
        
```
